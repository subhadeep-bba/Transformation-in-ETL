1. ANS - Data transformation in ETL is the central phase where raw, inconsistent data from multiple sources is cleaned, structured, and converted into a standardized, analysis-ready format before it is loaded into a target destination like a data warehouse.

why it is important:-

1. Ensures Data Quality and Accuracy: Raw data often contains errors, duplicates, or corrupted values. Transformation eliminates these flaws so teams do not base decisions on faulty information.

2. Enables Seamless Integration: Data pulled from different systems (like CRM tools, spreadsheets, and SQL databases) uses different structures and naming conventions. Transformation unifies them into a single, cohesive schema.

3. Optimizes for Analytics and BI: Business intelligence tools and dashboards require clean, neatly organized tables to run queries quickly and generate accurate visualizations.

4. Improves Security and Compliance: Transforming data allows engineers to mask, encrypt, or filter out sensitive personal information before it reaches permanent storage

3. ANS - In ETL and data preprocessing, Normalization and Standardization are techniques used to scale numerical data so that different features can be compared easily and machine learning algorithms can work more effectively.

Normalization converts data into a specific range, usually 0 to 1, whereas Standardization converts data into a distribution with a mean of 0 and a standard deviation of 1.

4. ANS - (i)Mean or Median Imputation (ii)Removing Records with Missing Age

When they should be used:-

(i) Use mean imputation when the Age data is approximately normally distributed and does not contain extreme values.
(ii) Use median imputation when the data contains outliers or is skewed, because the median is less affected by extreme values.
(iii) Use this technique when only a small number of records have missing Age values.
(iv)Use it when Age is essential for the analysis and removing those records will not significantly affect the dataset.
(v)Avoid it when many records have missing values, as it may lead to loss of important information.

6. ANS - One-Hot Encoding is a data transformation technique used in ETL and machine learning to convert categorical data into numerical values (0 and 1).
It creates a separate binary column for each category. A value of 1 indicates that the category is present, while 0 indicates that it is absent.

Example

Suppose we have a dataset with the following Color categories:

["Red", "Blue", "Green", "Red"]

After applying One-Hot Encoding:

Color Red Blue Green Red
red    1    0    0    0
blue   0    1    0    0
green  0    0    1    0
red    1    0    0    0

7. ANS - In ETL (Extract, Transform, Load), Data Integration and Data Mapping are two important processes used to combine and organize data from different sources.

Difference :-
1. Data Integration

Data Integration is the process of combining data from multiple sources into a single system or destination so that it can be used for analysis and reporting.

2. Data Mapping

Data Mapping is the process of defining how data fields from a source system correspond to fields in the target system.
It specifies which source column should go into which target column and how the data should be transformed.
